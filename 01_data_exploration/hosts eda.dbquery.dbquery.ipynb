```sql
---hosts table
use samples.wanderbricks;

show tables;

select *from hosts;

select count(distinct host_id)as total_hosts from hosts;

---2.Country wise host
select count(distinct country)as total_countries from hosts;

---3.Total host by country
select country,count(distinct host_id)as total_hosts from hosts group by country order by total_hosts desc limit 10;

----4.Verified vs non-verified
select is_verified,count(distinct host_id)as total_hosts from hosts group by is_verified order by total_hosts desc;

--4.Active vs non-active
select is_active,count(distinct host_id)as total_hosts from hosts group by is_active order by total_hosts desc;

--5.by host rating
select rating,count(distinct host_id)as total_hosts from hosts group by rating order by total_hosts desc;

---6.when did host joined the platform
select joined_at,count(distinct host_id)as total_hosts from hosts group by joined_at order by total_hosts desc limit 10;

--7.How many properties does each host manage
select host_id,count(distinct property_id)as total_properties from properties group by host_id order by total_properties desc limit 10;

select *from property_amenities limit 100;
select *from amenities;

select count(*) from amenities;

---1.total aminities.
select count(distinct amenity_id)as total_aminities from amenities;

---2.what aminities category exists.
select distinct category from amenities;

---3.most common amentities
select name,count(*)as total_aminities from amenities group by name order by total_aminities desc limit 10;

---4.most common amentities category
select category,count(*)as total_aminities from amenities group by category order by total_aminities desc limit 10;

---5.how many amentities does each property have
select property_id,count(distinct amenity_id)as total_aminities from property_amenities group by property_id order by total_aminities desc limit 10;

---6.Avg amentities per property
select round(avg(count(distinct amenity_id)),2)as avg_aminities from property_amenities group by property_id

---7.most common amentities wihin each country
select country,name,count(*)as total_aminities from property_amenities pa join properties p on pa.property_id=p.property_id join hosts h on p.host_id=h.host_id join amenities a on pa.amenity_id=a.amenity_id group by country,name order by total_aminities desc limit 10;

---8.avg  property price by amentities category
select category,round(avg(price),2)as avg_price from property_amenities pa join properties p on pa.property_id=p.property_id join amenities a on pa.amenity_id=a.amenity_id group by category order by avg_price desc; 

```
