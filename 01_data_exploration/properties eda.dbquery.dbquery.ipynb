```sql
---properties table
use samples.wanderbricks;

show tables;

select *from properties LIMIT 100;

describe properties;

----1. Total Properties
 SELECT COUNT(*) AS total_properties
FROM properties;

----2. Preview Data
SELECT *
FROM properties
LIMIT 20;

-----3. Check Duplicate Property IDs
SELECT property_id,
COUNT(*) AS total
FROM properties
GROUP BY property_id
HAVING COUNT(*)>1;

----4. Check Null Values
SELECT
COUNT(*) total_rows,
COUNT(property_id) property_id,
COUNT(host_id) host_id,
COUNT(destination_id) destination_id,
COUNT(title) title,
COUNT(base_price) base_price,
COUNT(property_type) property_type,
COUNT(max_guests) max_guests,
COUNT(bedrooms) bedrooms,
COUNT(bathrooms) bathrooms,
COUNT(created_at) created_at
FROM properties;

----Section 2: Basic Business Analysis
----5. Total Number of Property Types
SELECT DISTINCT property_type
FROM properties;

----6. Number of Properties by Type
SELECT
property_type,
COUNT(*) total_properties
FROM properties
GROUP BY property_type
ORDER BY total_properties DESC;

----7. Average Base Price
SELECT
ROUND(AVG(base_price),2) average_price
FROM properties;

----8. Minimum and Maximum Property Price
SELECT
MIN(base_price) lowest_price,
MAX(base_price) highest_price
FROM properties;

----9. Top 10 Most Expensive Properties
SELECT
property_id,
title,
property_type,
base_price
FRom properties
ORDER BY base_price DESC
LIMIT 10;

----10. Top 10 Cheapest Properties
SELECT
property_id,
title,
property_type,
base_price
FROM properties
ORDER BY base_price
LIMIT 10;

----Section 3: Capacity Analysis
----11. Average Guest Capacity
SELECT
ROUND(AVG(max_guests),2) average_guest_capacity
FROM properties;

----12. Property Type with Highest Guest Capacity
SELECT
property_type,
ROUND(AVG(max_guests),2) avg_capacity
FROM properties
GROUP BY property_type
ORDER BY avg_capacity DESC;

----13. Properties That Can Accommodate More Than 6 Guests
SELECT
property_id,
title,
max_guests
FROM properties
WHERE max_guests>6;

----14. Distribution of Guest Capacity
SELECT
max_guests,
COUNT(*) total_properties
FROM properties
GROUP BY max_guests
ORDER BY max_guests;

----Section 4: Bedroom Analysis
----15. Average Bedrooms
SELECT
ROUND(AVG(bedrooms),2)
FROM properties;

----16. Average Bedrooms by Property Type
SELECT
property_type,
ROUND(AVG(bedrooms),2) average_bedrooms
FROM properties
GROUP BY property_type;

----17. Largest Properties
SELECT
title,
bedrooms,
bathrooms,
max_guests
FROM properties
ORDER BY bedrooms DESC,
bathrooms DESC;

----Section 5: Price Analysis
----18. Average Price by Property Type
SELECT
property_type,
ROUND(AVG(base_price),2) average_price
FROM properties
GROUP BY property_type
ORDER BY average_price DESC;

----19. Premium Properties
SELECT
property_id,
title,
base_price
FROM properties
WHERE base_price >
(
SELECT AVG(base_price)
FROM properties
);

----20. Budget Properties
SELECT
property_id,
title,
base_price
FROM properties
WHERE base_price <
(
SELECT AVG(base_price)
FROM properties
);
----Section 6: Host Analysis
----21. Hosts with Most Properties
SELECT
host_id,
COUNT(*) total_properties
FROM properties
GROUP BY host_id
ORDER BY total_properties DESC;

----22. Hosts Owning More Than One Property
SELECT
host_id,
COUNT(*) total_properties
FROM properties
GROUP BY host_id
HAVING COUNT(*)>1
ORDER BY total_properties DESC;

 Section 7: Time Analysis

----23. Properties Added Every Year
SELECT
YEAR(created_at) year,
COUNT(*) total_properties
FROM properties
GROUP BY YEAR(created_at)
ORDER BY year;
```
