```sql
---reviews table
use samples.wanderbricks;

show tables;

select *from reviews limit 5;

---total reviews
select count(*) from reviews;

---2. Total Unique Properties Reviewed
SELECT COUNT(DISTINCT property_id) AS reviewed_properties
FROM reviews;

---3.total unique reviewers
SELECT COUNT(DISTINCT user_id) AS unique_reviewers
FROM reviews;

---4. Deleted Reviews
SELECT
is_deleted,
COUNT(*) AS total_reviews
FROM reviews
GROUP BY is_deleted;

---5.Average rating
SELECT AVG(rating) AS avg_rating
FROM reviews;

---6. Top 5 properties with highest rating
SELECT
property_id,
AVG(rating) As avg_rating
FROM reviews
GROUP BY property_id
ORDER BY avg_rating DESC
LIMIT 5;

---7. Highest & Lowest Rating
SELECT
MAX(rating) AS highest_rating,
MIN(rating) AS lowest_rating
FROM reviews;

---8. Median Rating
SELECT
percentile_approx(rating,0.5) AS median_rating
FROM reviews;

---8. Rating Distribution
SELECT
rating,
COUNT(*) AS total_reviews
FROM reviews
GROUP BY rating
ORDER BY rating DESC;

---9. Top Rated Properties
SELECT
property_id,
ROUND(AVG(rating),2) AS avg_rating,
COUNT(*) AS total_reviews
FROM reviews
GROUP BY property_id
HAVING COUNT(*)>=5
ORDER BY avg_rating DESC
LIMIT 10;

---10. Lowest Rated Properties
SELECT
property_id,
ROUND(AVG(rating),2) AS avg_rating,
COUNT(*) AS total_reviews
FROM reviews
GROUP BY property_id
HAVING COUNT(*)>=5
ORDER BY avg_rating
LIMIT 10;

---11. Most Reviewed Properties
SELECT
property_id,
COUNT(*) AS review_count
FROM reviews
GROUP BY property_id
ORDER BY review_count DESC
LIMIT 10;

---12. Most Active Reviewers
SELECT
user_id,
COUNT(*) AS total_reviews
FROM reviews
GROUP BY user_id
ORDER BY total_reviews DESC
LIMIT 10;

---13. Monthly Reviews
SELECT
DATE_FORMAT(created_at,'yyyy-MM') AS month,
COUNT(*) AS total_reviews
FROM reviews
GROUP BY DATE_FORMAT(created_at,'yyyy-MM')
ORDER BY month;

---14. Average Monthly Rating
SELECT
DATE_FORMAT(created_at,'yyyy-MM') AS month,
ROUND(AVG(rating),2) AS average_rating
FROM reviews
GROUP BY DATE_FORMAT(created_at,'yyyy-MM')
ORDER BY month;

---15. Reviews Updated After Creation
SELECT
COUNT(*) AS updated_reviews
FROM reviews
WHERE updated_at > created_at;

---16. Longest Reviews
SELECT
review_id,
property_id,
LENGTH(comment) AS comment_length
FROM reviews
ORDER BY comment_length DESC
LIMIT 10;

---17. Average Review Length
SELECT
ROUND(AVG(LENGTH(comment)),2) AS avg_comment_length
FROM reviews;
```
