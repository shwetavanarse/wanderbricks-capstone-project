```sql
use samples.wanderbricks;
show tables;
----exploring users table
select * from users limit 10;
describe users;
----count of users
select count(*) as total_rows,count(distinct user_id) as unique_users from users;

---users table

--1.contains information about wanderbricks user
---2.one row represent one user
---3.columns tells us what information we know about each user
---4.There are 100000 unique users in table
---5.there are null values in email and company_name column
---6.user can be either individual or business
---7.there are individual users and business

---what is the distribution of users by country?
select country, count(*) as run_users from users  group by country order by run_users desc;

---distribution by user type
select user_type, count(*) as run_users from users group by user_type order by run_users desc;

---any missing value
select * from users where user_type is null or country is null or name is null or email is null or company_name is null or is_business is null; 
---what does our user base look like?
select country,user_type,count(*)
from users
group by country,user_type
order by country,user_type;

---when did users join the platform(earliest sign up,latest signup,users joining by month/year)
select min(created_at) as earliest_signup,max(created_at) as latest_signup 
from users;
select year(created_at) as signup_year,month(created_at) as signup_month,count(*) as total_users
from users
group by year(created_at),month(created_at)
order by signup_year,signup_month;
```
